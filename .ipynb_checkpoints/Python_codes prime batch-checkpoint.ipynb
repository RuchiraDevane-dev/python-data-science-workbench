```python
# if,elif,else
username = input("enter username :")
password = input("enter password :")
if (username == "admin" and password == "Pass"):
    print(" u can login")
elif (username != "admin" and password == "Pass"):
    print("username is incorrect ")
else:
    print("password is incorrect ")
```
