```python
#new testing on emp data
#read cs file
df = pd.read_csv("employee_data.csv")
#basic dataframe checking
#print all data
df
#first 5 rows
df.head()
#last five rows
df.tail()
#shows row and column
df.shape
#show all columns name
df.columns
#Shows datatype, null values, memory usage.
df.info()
#Gives statistics for numeric columns.
df.describe()
#Gives summary for all columns.
df.describe(include="all")


#checking null values
#check missing cvalues as T/F
df.isnull()
#Counts missing values column-wise.
df.isnull().sum()
#Shows missing value percentage.
df.isnull().mean() * 100



#handle null values
#Removes rows having any null value.
df.dropna()

#Fill all null values with specific value
df = df.fillna("NAN")
df




#Check duplicate values
#Returns True for duplicate rows.
df.duplicated()
#Counts duplicate rows.
df.duplicated().sum()
#drop duplicated rows
df = df.drop_duplicates()
df


#selects first row by index position.
df.iloc[0]
# #Selects first 5 rows.
df.iloc[0:5]
# #Selects row by label index.
df.loc[0]


df.to_json("cleaned_employee_data.json")
df_json = pd.read_json("cleaned_employee_data.json")
print(df)


import json

with open("cleaned_employee_data.json", "r") as file:
    data = json.load(file)

print(json.dumps(data, indent=4))
```
